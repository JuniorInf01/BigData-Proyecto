```python
from pyspark.sql import functions as F
p=spark.read.parquet(ruta_hdfs('processed','tabla_analitica_mensual'))
print('Esquema definitivo:');p.printSchema()
print('Conteo total:',p.count())
print('Clave única:',p.select('fecha_mes').distinct().count())
print('Rango de meses:');p.agg(F.min('fecha_mes'),F.max('fecha_mes')).show()
print('Diciembre 2018 sistema desde SBS:')
p.filter(F.col('fecha_mes')=='2018-12-01').select('fecha_mes','rent_real_sistema_12m','fuente_sistema').show()
print('Mes final sin etiqueta de futuro:')
p.orderBy(F.desc('fecha_mes')).select('fecha_mes','objetivo_bajo_rendimiento_mes_siguiente').show(2)
```
